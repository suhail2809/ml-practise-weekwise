```python
plt.figure(figsize=(10,10))

x=plt.subplot(2,2,1)
x= np.linspace(1/np.e**2, np.e**3)
y = np.log(x)

plt.plot(x,y)
plt.title("y = ln(x)")
x = plt.subplot(2,2,2)
x = np.linspace(0,3)
y = np.e**x
plt.plot(x,y)
plt.title("$y = e^{x}$");

# plot-3 :-
# y = sin(x), x∈(0,2π)
x = plt.subplot(2,2,3) ;
x = np.linspace(0,2*np.pi)
y = np.sin(x)
plt.plot(x,y)
plt.title("$y = sin(x)$");

#plot-4 :-
# y = 1/x, x∈(1,5)
x = plt.subplot(2,2,4)
x = np.linspace(1,5)
y = 1/(x) ;
plt.plot(x,y) ;
plt.title("$y = 1/x$");
```

```python
# y = ln(x)
plt.figure(figsize=(10,8))
#is method se tum jo subplot bana rhe ho uska size har graph ka set kr sakte ho
#nhi krne pe dusre graph ka title aur graph mix ho jata hai

plt.subplot(2,2,1) #----> 2,2 ka matlab 2 by2 ke form me subplot ko divide kr
# dega aur jo last ka value hai (1)wo index batata hai ki 2x2 ke subplot me wo graph kidhar hoga
x=np.linspace(1/(np.e**2),np.e**3)
y=np.log(x)
plt.plot(x,y)
plt.title("$y=$ln(x)");


plt.subplot(2,2,2) #----> ye subplot gaya dusre index pe
x = np.linspace(0,3)
y= (np.e)**x
plt.plot(x,y)
plt.title("$y=$e^x");


plt.subplot(2,2,3) #-----> ye 3rd index joki 2x2 ka 3rd index 1st index ke niche hoga
x=np.linspace(0,2*np.pi)
y= np.sin(x)  #--> sin fn ko kaise multiply karana hai to check numpy documentation!!
plt.plot(x,y)
plt.title("$y=$sin(x)");


plt.subplot(2,2,4)  #4th index
x=np.linspace(1,5)
y= 1/(x)
plt.plot(x,y)
plt.title("$y=$1/x");
```
